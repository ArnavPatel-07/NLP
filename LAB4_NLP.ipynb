# Experiment 4: Term Frequency and Named Entity Recognition

## Experiment 4.1: Term Frequency and NER using a toolkit (spaCy)
Read text from file, compute term frequency, save as CSV (Term, Frequency), show top 10 terms, and perform Named Entity Recognition.

```python
!pip install spacy -q
!python -m spacy download en_core_web_sm -q

import spacy
import csv
from collections import Counter

nlp = spacy.load("en_core_web_sm")

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as f:
    text = f.read()

doc = nlp(text)

# Term frequency (alphabetic tokens only, lowercased)
tokens = [token.text.lower() for token in doc if token.is_alpha]
freq = Counter(tokens)

with open("term_frequency_4_1.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Term", "Frequency"])
    for term, count in freq.items():
        writer.writerow([term, count])

print("Top 10 most frequent terms:")
for term, count in freq.most_common(10):
    print(f"{term}: {count}")

print("\nNamed Entities:")
for ent in doc.ents:
    print(f"{ent.text} -> {ent.label_}")
```

## Experiment 4.2: Term Frequency without using any toolkit
Read text from file, manually clean and tokenize it, compute term frequency, save as CSV, and show the top 10 terms.

```python
import string
import csv
from collections import Counter

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Lowercase and remove punctuation
text_clean = text.lower()
for ch in string.punctuation:
    text_clean = text_clean.replace(ch, " ")

tokens = text_clean.split()
freq = Counter(tokens)

with open("term_frequency_4_2.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Term", "Frequency"])
    for term, count in freq.items():
        writer.writerow([term, count])

print("Top 10 most frequent terms:")
for term, count in freq.most_common(10):
    print(f"{term}: {count}")
```

## Experiment 4.3: TF-IDF from scratch (no NLP toolkit)
Compute TF, DF, IDF, and TF-IDF manually for three given documents, and display the top 10 TF-IDF terms per document.

```python
import string
import math
from collections import Counter

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

def clean_and_tokenize(text):
    text = text.lower()
    for ch in string.punctuation:
        text = text.replace(ch, " ")
    return text.split()

tokenized_docs = [clean_and_tokenize(doc) for doc in documents]
N = len(tokenized_docs)

# Term Frequency per document
tf_list = []
for tokens in tokenized_docs:
    counts = Counter(tokens)
    total_terms = len(tokens)
    tf = {term: count / total_terms for term, count in counts.items()}
    tf_list.append(tf)

# Document Frequency across all documents
all_terms = set(term for tokens in tokenized_docs for term in tokens)
df = {term: sum(1 for tokens in tokenized_docs if term in tokens) for term in all_terms}

# Inverse Document Frequency
idf = {term: math.log(N / df[term]) for term in all_terms}

# TF-IDF per document
tfidf_list = []
for tf in tf_list:
    tfidf = {term: tf[term] * idf[term] for term in tf}
    tfidf_list.append(tfidf)

# Display Term, TF, DF, IDF, TF-IDF for each document
for i, (tf, tfidf) in enumerate(zip(tf_list, tfidf_list), start=1):
    print(f"\nDocument {i}:")
    print(f"{'Term':<15}{'TF':<10}{'DF':<5}{'IDF':<10}{'TF-IDF':<10}")
    for term in tf:
        print(f"{term:<15}{tf[term]:<10.4f}{df[term]:<5}{idf[term]:<10.4f}{tfidf[term]:<10.4f}")

    print(f"\nTop 10 TF-IDF terms in Document {i}:")
    top_terms = sorted(tfidf.items(), key=lambda x: x[1], reverse=True)[:10]
    for term, score in top_terms:
        print(f"{term}: {score:.4f}")
```
