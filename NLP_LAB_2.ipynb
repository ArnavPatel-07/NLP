## Experiment 2.1: Basic Text Preprocessing

**Objective:** Write a Python program to perform basic text preprocessing on a given piece of text.

**Tasks:**

1. **Convert text to lowercase**
   - Also display how many uppercase characters were present in the original text.

2. **Remove punctuation**
   - Also display how many different types of punctuation marks were present.

3. **Remove numbers**
   - Also display how many numbers (digits) were present.

4. **Remove extra whitespace**
   - Also display how much extra whitespace was present.

5. **Display the cleaned text**
   - Show the final preprocessed output after all the above steps.



```python
# Cell 1: Load text data
text = open("2.1_text_data.txt", "r", encoding="utf-8").read()
print("Original:", text)
```

```python
# Cell 2: Count uppercase letters
uppercase_count = sum(1 for ch in text if ch.isupper())
print("Uppercase Count:", uppercase_count)
```

```python
# Cell 3: Convert to lowercase
text = text.lower()
print("Lowercase Text:", text)
```

```python
# Cell 4: Count punctuation types
import string
punct_types = set(ch for ch in text if ch in string.punctuation)
print("Punctuation Types:", punct_types)
print("Number of Different Punctuation Types:", len(punct_types))
```

```python
# Cell 5: Remove punctuation
text = "".join(ch for ch in text if ch not in string.punctuation)
print("Text without Punctuation:", text)
```

```python
# Cell 6: Count numbers
numbers = [ch for ch in text if ch.isdigit()]
print("Numbers Found:", numbers)
print("Count of Numbers:", len(numbers))
```

## Experiment 2.2: Word and Sentence Tokenization

**Objective:** Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

**Tasks:**

1. **Word Tokenization** — Split the text into individual words (tokens).
2. **Sentence Tokenization** — Split the text into individual sentences.
3. Display words and sentences separately.

**Note:** This must be implemented in **three separate ways**:
   - (a) Using **NLTK**
   - (b) Using **spaCy**
   - (c) **Without** using NLTK or spaCy (pure Python / manual logic)

---


```python
# Cell 1: Load text data
text = open("2.2_tokenization_data.txt", "r", encoding="utf-8").read()
print("Original Text:", text)
```

(a) Using NLTK

```python
# Cell 2: Sentence tokenization using NLTK
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(text)
print("Sentences:", sentences)
print("Number of Sentences:", len(sentences))
```

```python
# Cell 3: Word tokenization using NLTK
from nltk.tokenize import word_tokenize

words = word_tokenize(text)
print("Words:", words)
print("Number of Words:", len(words))
```

(b) Using spaCy

```python
# Cell 4: Load spaCy model
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
```

```python
# Cell 5: Sentence tokenization using spaCy
sentences_spacy = [sent.text for sent in doc.sents]
print("Sentences:", sentences_spacy)
print("Number of Sentences:", len(sentences_spacy))
```

```python
# Cell 6: Word tokenization using spaCy
words_spacy = [token.text for token in doc]
print("Words:", words_spacy)
print("Number of Words:", len(words_spacy))
```

(c) Without using NLTK/spaCy

```python
# Cell 7: Sentence tokenization manually
import re

sentences_manual = re.split(r'(?<=[.!?])\s+', text.strip())
print("Sentences:", sentences_manual)
print("Number of Sentences:", len(sentences_manual))
```

```python
# Cell 8: Word tokenization manually
words_manual = text.split()
print("Words:", words_manual)
print("Number of Words:", len(words_manual))
```

## Experiment 2.3: Stop Word Removal

**Objective:** Write a Python program to tokenize a text and remove commonly occurring stop words such as *is, the, a, an, in, of, and, to,* etc. Display the original and filtered tokens.

**Tasks:**

1. **Tokenize** the given text into words.
2. **Identify and remove stop words** from the tokenized text.
3. **Display the original tokens** (before removal).
4. **Display the filtered tokens** (after removal).
5. **Display the stop words separately** (the ones that were removed).



```python
# Cell 1: Load text data
text = open("2.3_clean_data.txt", "r", encoding="utf-8").read()
print("Original Text:", text)
```

```python
# Cell 2: Tokenize text
import re

tokens = re.findall(r'\b\w+\b', text.lower())
print("Original Tokens:", tokens)
print("Number of Tokens:", len(tokens))
```

```python
# Cell 3: Define stop words
stop_words = {"is", "the", "a", "an", "in", "of", "and", "to", "for", "on",
              "with", "from", "by", "this", "that", "it", "as", "such"}
print("Stop Words List:", stop_words)
```

```python
# Cell 4: Separate stop words found in text
stop_words_found = [word for word in tokens if word in stop_words]
print("Stop Words Found in Text:", stop_words_found)
print("Number of Stop Words:", len(stop_words_found))
```

```python
# Cell 5: Remove stop words
filtered_tokens = [word for word in tokens if word not in stop_words]
print("Filtered Tokens:", filtered_tokens)
print("Number of Filtered Tokens:", len(filtered_tokens))
```
