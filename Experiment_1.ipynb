# Experiment-1: Data Analysis

**Objective:** Perform data analysis and visualization on an employee dataset using Python (Pandas, Matplotlib, Seaborn).

**Dataset:** employee_information_100.csv  
**Columns:** Employee_ID, Name, Department, Age, Gender, Salary, Experience_Years

---

### Experiment-1.1: Average salary of each department (Bar Chart)
### Experiment-1.2: Number of employees in each department
### Experiment-1.3: Percentage of male and female employees (Pie Chart)
### Experiment-1.4: Salary distribution (Histogram)
### Experiment-1.5: Relationship between experience and salary (Scatter Plot)
### Experiment-1.6: Top 10 employees with highest salary
### Experiment-1.7: Highest salary in every department
### Experiment-1.8: Employees with salary greater than overall average
### Experiment-1.9: Average years of experience in each department
### Experiment-1.10: Distribution of employee ages (Histogram)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# If running on Google Colab, upload the file:
from google.colab import files
uploaded = files.upload()   # choose employee_information_100.csv

df = pd.read_csv("employee_information_100.csv")
df.head()
```

```python
avg_salary_dept = df.groupby("Department")["Salary"].mean()

plt.figure(figsize=(8,5))
avg_salary_dept.plot(kind="bar", color="skyblue")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
emp_count_dept = df["Department"].value_counts()

plt.figure(figsize=(8,5))
emp_count_dept.plot(kind="bar", color="orange")
plt.title("Number of Employees per Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(emp_count_dept)
```

```python
gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender_counts, labels=gender_counts.index, autopct="%1.1f%%",
        colors=["lightblue", "lightpink"], startangle=90)
plt.title("Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Salary"], bins=15, color="green", edgecolor="black")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Experience_Years"], df["Salary"], color="purple", alpha=0.6)
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.tight_layout()
plt.show()
```

```python
top10_salary = df.sort_values(by="Salary", ascending=False).head(10)
top10_salary[["Employee_ID", "Name", "Department", "Salary"]]
```

```python
highest_salary_dept = df.groupby("Department")["Salary"].max()
highest_salary_dept
```

```python
overall_avg_salary = df["Salary"].mean()
above_avg_emp = df[df["Salary"] > overall_avg_salary]

print(f"Overall Average Salary: {overall_avg_salary:.2f}")
above_avg_emp[["Employee_ID", "Name", "Department", "Salary"]]
```

```python
avg_exp_dept = df.groupby("Department")["Experience_Years"].mean()

plt.figure(figsize=(8,5))
avg_exp_dept.plot(kind="bar", color="teal")
plt.title("Average Experience by Department")
plt.xlabel("Department")
plt.ylabel("Average Experience (Years)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(avg_exp_dept)
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=15, color="coral", edgecolor="black")
plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
